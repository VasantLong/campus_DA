```python
from pyecharts_exporter import display_chart
from pyecharts.charts import Map, Geo
from pyecharts import options as opts
from pyecharts.globals import ChartType, SymbolType
import pandas as pd
import chardet
```

## 1
地图Map（）― 分省GDP地图可视化
请根据分省GDP年度数据.csv数据，绘制2019年分省GDP分布地图。效果如下图所示。
 


```python
# 1. 读取 CSV 数据
file_path = './data/分省年度数据.csv'

# 获取编码
with open(file_path, 'rb') as f:
    encoding = chardet.detect(f.read())['encoding']

# 使用编码读取
df = pd.read_csv(file_path, encoding=encoding)

# 2. 提取2019年的数据
df_2019 = df[['地区', '2019年']].copy()

# 3. 数据清洗：将地区名称转换为地图识别的标准名称
# 定义地区名称映射
name_mapping = {
    '西藏省': '西藏自治区',
    '新疆省': '新疆维吾尔自治区'
}

# 应用名称映射
df_2019['地区'] = df_2019['地区'].replace(name_mapping)

# 删除可能的空值（如果有映射不成功的）
df_2019 = df_2019.dropna()

# 4. 准备地图数据
data_list = df_2019.values.tolist()
```

```python
map_chart = (
    Map(init_opts=opts.InitOpts(width="1200px", height="800px", theme="light"))
    .add(
        series_name="GDP (亿元)",
        data_pair=data_list,
        maptype="china",
        is_map_symbol_show=False,
        label_opts=opts.LabelOpts(is_show=True, formatter="{b}"),
        itemstyle_opts=opts.ItemStyleOpts(
            border_color="#fff",
            border_width=0.5,
        ),
    )
    .set_global_opts(
        title_opts=opts.TitleOpts(
            title="2019年中国各省份GDP分布地图",
            subtitle="单位：亿元",
            pos_left="center",
        ),
        visualmap_opts=opts.VisualMapOpts(
            min_=df_2019['2019年'].min(),
            max_=df_2019['2019年'].max(),
            is_calculable=True,
            # 蓝色系渐变配置
            #range_color=["#c6e2ff", "#5ba3e6", "#2171b5", "#084594"],  # 浅蓝到深蓝
            # 或者使用更丰富的蓝色渐变
            range_color=["#e0f3f8", "#abd9e9", "#74add1", "#4575b4", "#313695"],
        ),
        tooltip_opts=opts.TooltipOpts(
            is_show=True,
            trigger="item",
            formatter="{b}<br/>GDP: {c} 亿元",
        ),
    )
)

display_chart(map_chart)
```

请根据你旅游去过的城市及对该城市的推荐值，绘制出“♡♡♡ 往后余生，踏遍山河 ♡♡♡”图。图类型可以选择动态涟漪散点图显示各城市分布，也可以绘制去旅游城市的线路。

```python
from pyecharts.commons.utils import JsCode

cities_data = [
    ("北京", 95),
    ("天津", 85),
    ("上海", 92),
    ("淄博", 82),
    ("青岛", 90),
    ("烟台", 83),
    ("苏州", 96),
    ("杭州", 94),
    ("无锡", 80),
    ("常州", 75),
    ("镇江", 78),
    ("连云港", 76),
    ("长沙", 88),
    ("张家界", 93),
    ("晋中", 84),
    ("太原", 80),
    ("廊坊", 72),
]

travel_routes = [
    # 京津冀线
    ("天津", "廊坊"),
    ("廊坊", "北京"),
    # 山东线
    ("淄博", "烟台"),
    ("烟台", "青岛"),
    # 苏南线
    ("连云港", "镇江"),
    ("镇江", "常州"),
    ("常州", "无锡"),
    ("无锡", "苏州"),
    ("苏州", "上海"),
    # 苏杭线
    ("上海", "杭州"),
    # 山西线
    ("太原", "晋中"),
    # 湖南线
    ("长沙", "张家界"),
    # 跨区域连线
    ("天津", "淄博"),
    ("天津", "晋中"),
    ("天津", "长沙"),
    ("天津", "连云港"),
]

# 旅游线路（按区域分组，模拟你的旅行轨迹）
# 线路1：京津冀线 - 北京 → 廊坊 → 天津
# 线路2：山东线 - 淄博 → 青岛 → 烟台
# 线路3：苏南线 - 连云港 → 镇江 → 常州 → 无锡 → 苏州 → 上海
# 线路4：苏杭线 - 上海 → 杭州
# 线路5：山西线 - 太原 → 晋中
# 线路6：湖南线 - 长沙 → 张家界

```

```python

c = (
    Geo(init_opts=opts.InitOpts(width="1400px", height="900px"))
    .add_schema(
        maptype="china", 
        itemstyle_opts=opts.ItemStyleOpts(
            color="#1a1a2e",           # 省份内部填充色（深色背景）
            border_color="#4a4a6a",    # 🔥 省份分界线颜色（灰紫色）
            border_width=1.2,          # 🔥 省份分界线宽度
),
    )
    # 旅游路线 - 动态箭头连线
    .add(
        "旅游路线",
        travel_routes,
        type_=ChartType.LINES,
        effect_opts=opts.EffectOpts(
            symbol=SymbolType.ARROW, 
            symbol_size=10, 
            color="#00ffff",       # 箭头颜色改为亮青色
            period=3,
            trail_length=0.3,      # 拖尾长度
        ),
        linestyle_opts=opts.LineStyleOpts(
            curve=0.2, 
            color="#00ffff",        # 线条颜色改为亮青色
            width=2.5,              # 加粗线条
            opacity=0.9,            # 提高不透明度
            type_="solid",          # 实线
        ),
        label_opts=opts.LabelOpts(is_show=False),
    )
    # 涟漪散点 - 城市标注
    .add(
        "推荐值",
        cities_data,
        type_=ChartType.EFFECT_SCATTER,
        symbol_size=12,
        effect_opts=opts.EffectOpts(
            scale=3.5, 
            brush_type="stroke",
            period=6,
        ),
        label_opts=opts.LabelOpts(
            is_show=True,
            position="right",
            formatter=JsCode(
                """
                function(params) {
                    return params.name + '\\n推荐: ' + params.value[2] + '分';
                }
                """
            ),
            font_size=10,
            color="#eee",
        ),
        itemstyle_opts=opts.ItemStyleOpts(color="#ff6b81"),
    )
    .set_series_opts()
    .set_global_opts(
        title_opts=opts.TitleOpts(
            title="♡♡♡ 往后余生，踏遍山河 ♡♡♡",
            subtitle="17个城市 · 6条旅行线路",
            pos_left="center",
            title_textstyle_opts=opts.TextStyleOpts(
                font_size=32, 
                color="#ff6b81", 
                font_family="KaiTi, STKaiti",
                font_weight="bold",
            ),
            subtitle_textstyle_opts=opts.TextStyleOpts(
                font_size=16, 
                color="#00d2ff",
                font_family="KaiTi, STKaiti",
            ),
        ),
        visualmap_opts=opts.VisualMapOpts(
            is_show=True,
            min_=70,
            max_=100,
            range_color=["#a2d5c6", "#f4a261", "#e76f51"],
            pos_left="left",
            pos_bottom="bottom",
            textstyle_opts=opts.TextStyleOpts(color="#eee"),
        ),
        tooltip_opts=opts.TooltipOpts(
            trigger="item",
            formatter=JsCode(
                """
                function(params) {
                    if (params.seriesType === 'effectScatter') {
                        return params.name + '<br/>推荐值: ' + params.value[2] + ' 分';
                    } else if (params.seriesType === 'lines') {
                        return '路线: ' + params.name;
                    }
                    return params.name;
                }
                """
            ),
        ),
    )
)


display_chart(c)
```
