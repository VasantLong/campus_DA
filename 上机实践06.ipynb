```python
from pyecharts_exporter import display_chart
from pyecharts.charts import Line, Bar
from pyecharts import options as opts
import pandas as pd
import chardet
```

## 1

```python
file_path = './data/广播电视情况.csv'

# 获取编码
with open(file_path, 'rb') as f:
    encoding = chardet.detect(f.read())['encoding']

# 使用编码读取
df = pd.read_csv(file_path, encoding=encoding)

# 查看数据结构
print(df.head())
print(df.columns.tolist())
```

```python
# 提取2015-2019年的数据
# 根据数据，行是指标，列是年份，需要转置
df_years = df.set_index('指标').T

# 提取2015-2019年的全国有线广播电视用户数和农村有线广播电视用户数
years = ['2015年', '2016年', '2017年', '2018年', '2019年']
national_users = []
rural_users = []

for year in years:
    national_users.append(df_years.loc[year, '全国有线广播电视用户数(万户)'])
    rural_users.append(df_years.loc[year, '农村有线广播电视用户数(万户)'])

print(f"年份: {years}")
print(f"全国用户数: {national_users}")
print(f"农村用户数: {rural_users}")

# 创建折线图
line = Line(init_opts=opts.InitOpts(width="800px", height="500px"))

# 添加x轴数据
line.add_xaxis(years)

# 添加全国有线广播电视用户数折线
line.add_yaxis(
    series_name="全国有线广播电视用户数",
    y_axis=national_users,
    label_opts=opts.LabelOpts(is_show=True, position="bottom"),  # 在底部显示标签
    linestyle_opts=opts.LineStyleOpts(width=2),
    symbol_size=8,
)

# 添加农村有线广播电视用户数折线
line.add_yaxis(
    series_name="农村有线广播电视用户数",
    y_axis=rural_users,
    label_opts=opts.LabelOpts(is_show=True, position="bottom"),  # 在底部显示标签
    linestyle_opts=opts.LineStyleOpts(width=2),
    symbol_size=8,
)

# 设置全局配置项
line.set_global_opts(
    title_opts=opts.TitleOpts(
        title="5年期间有线广播电视用户数情况",
        subtitle="单位：万户",
        title_link="",
        pos_left="center",  # 标题居中
    ),
    legend_opts=opts.LegendOpts(
        pos_top="10%",  # 图例距离顶部10%
        orient="horizontal",  # 水平布局
    ),
    xaxis_opts=opts.AxisOpts(
        name="年份",
        name_location="middle",
        name_gap=35,
    ),
    yaxis_opts=opts.AxisOpts(
        name="用户数",
        name_location="middle",
        name_gap=45,
    ),
    # 设置脚注
    #tooltip_opts=opts.TooltipOpts(trigger="axis"),#当鼠标悬停时，触发同一轴上所有系列的数据点提示
        graphic_opts=[
            opts.GraphicGroup(
                graphic_item=opts.GraphicItem(
                    left="center",
                    bottom="10px",
                    z=100
                ),
                children=[
                    opts.GraphicText(
                        graphic_item=opts.GraphicItem(
                            left="center",
                            bottom="20px"
                        ),
                        graphic_textstyle_opts=opts.GraphicTextStyleOpts(
                            text="数据来源：国家统计局",
                            font_size=12,
                            text_align="center"
                        )
                    )
                ]
            )
        ]
)


# 渲染图表
#line.render("有线广播电视用户数情况.html")

# 在Jupyter Notebook中显示（如果使用Notebook）
# line.render_notebook()

display_chart(line)
```

## 2

```python
# 文件路径
file_path = './data/人口年龄结构.csv'

# 获取编码
with open(file_path, 'rb') as f:
    encoding = chardet.detect(f.read())['encoding']

# 使用编码读取
df = pd.read_csv(file_path, encoding=encoding)

# 查看数据结构
print(df.head())
print(df.columns.tolist())
```

```python
# 提取2015-2019年的数据
df_years = df.set_index('指标').T

years = ['2015年', '2016年', '2017年', '2018年', '2019年']

# 提取柱状图数据（人口数，单位：万人）
total_population = []
population_0_14 = []
population_15_64 = []
population_65_above = []

# 提取折线图数据（抚养比，单位：%）
total_dependency = []
child_dependency = []
old_dependency = []

for year in years:
    total_population.append(df_years.loc[year, '年末总人口(万人)'])
    population_0_14.append(df_years.loc[year, '0-14岁人口(万人)'])
    population_15_64.append(df_years.loc[year, '15-64岁人口(万人)'])
    population_65_above.append(df_years.loc[year, '65岁及以上人口(万人)'])
    
    total_dependency.append(df_years.loc[year, '总抚养比(%)'])
    child_dependency.append(df_years.loc[year, '少儿抚养比(%)'])
    old_dependency.append(df_years.loc[year, '老年抚养比(%)'])

print(f"年份: {years}")
print(f"年末总人口: {total_population}")
print(f"总抚养比: {total_dependency}")

# 绘制柱状图（人口数）
bar = (
    Bar(init_opts=opts.InitOpts(width="1200px", height="600px"))
    .add_xaxis(years)
    .add_yaxis(
        "年末总人口", 
        total_population, 
        bar_width="15%",  # 减小柱子间隙，使柱子更细
        label_opts=opts.LabelOpts(is_show=True, position="top"),  # 数据放在柱子上方
        itemstyle_opts=opts.ItemStyleOpts(opacity=0.7)  # 设置透明度0.7
    )
    .add_yaxis(
        "0～14岁人口", 
        population_0_14, 
        bar_width="15%",
        label_opts=opts.LabelOpts(is_show=True, position="top")
    )
    .add_yaxis(
        "15～64岁人口", 
        population_15_64, 
        bar_width="15%",        
        label_opts=opts.LabelOpts(is_show=True, position="top")
    )
    .add_yaxis(
        "65岁及以上人口", 
        population_65_above, 
        bar_width="15%",        
        label_opts=opts.LabelOpts(is_show=True, position="top")
    )
    .set_global_opts(
        title_opts=opts.TitleOpts(
            title="5年期间年末人口数构成及抚养比情况",
            subtitle="单位：万人",
            #pos_left="center",
        ),
        legend_opts=opts.LegendOpts(
            selected_mode="multiple",  # 多选模式
            pos_left="center",  # 居中
            #pos_top="bottom",  # 图例放在底部
            pos_bottom="10px",  # 距离底部-6px
            orient="horizontal",
        ),
        xaxis_opts=opts.AxisOpts(
            name="年份",
            name_location="middle",
            name_gap=35,
        ),
        yaxis_opts=opts.AxisOpts(
            name="人口数（万人）",
            name_location="middle",
            name_gap=50,
        ),
        tooltip_opts=opts.TooltipOpts(trigger="axis"),
        # 添加脚注
        graphic_opts=opts.GraphicGroup(
            graphic_item=opts.GraphicItem(left='center', bottom='0%'),
            children=[
                opts.GraphicText(
                    graphic_textstyle_opts=opts.GraphicTextStyleOpts(
                        text='数据来源：国家统计局',
                        font_size=12,
                    )
                )
            ]
        )
    )
    # 在右边新增y坐标轴（用于抚养比）
    .extend_axis(
        yaxis=opts.AxisOpts(
            name="抚养比（%）",
            type_="value",
            position="right",
            axislabel_opts=opts.LabelOpts(formatter="{value} %"),
        )
    )
)

# 绘制折线图（抚养比）
line = (
    Line()
    .add_xaxis(xaxis_data=years)
    .add_yaxis(
        series_name="总抚养比",
        y_axis=total_dependency,
        yaxis_index=1,  # 绑定到右侧y轴
        label_opts=opts.LabelOpts(
            is_show=True,
            position="left",# 在左边显示标签
            formatter="{@[1]}%",# 格式为"数值%"
        ),
        linestyle_opts=opts.LineStyleOpts(width=2),
        symbol="circle",
        symbol_size=8,
    )
    .add_yaxis(
        series_name="少儿抚养比",
        y_axis=child_dependency,
        yaxis_index=1,
        label_opts=opts.LabelOpts(
            is_show=True,
            position="left",
            formatter="{@[1]}%",
        ),
        linestyle_opts=opts.LineStyleOpts(width=2, type_="dashed"),
        symbol="diamond",
        symbol_size=8,
    )
    .add_yaxis(
        series_name="老年抚养比",
        y_axis=old_dependency,
        yaxis_index=1,
        label_opts=opts.LabelOpts(
            is_show=True,
            position="left",
            formatter="{@[1]}%",
        ),
        linestyle_opts=opts.LineStyleOpts(width=2, type_="dotted"),
        symbol="triangle",
        symbol_size=8,
    )
)
```

```python

# 将折线图叠加到柱状图上
bar.overlap(line)

# 渲染图表
display_chart(bar)
```
