## Seaborn可视化实践——Iris实例

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df_iris = pd.read_csv('./data/iris.csv')
df_iris
```

## 【例1】绘制iris数据Petal_Width的分布图。

```python

#df_iris.head()
sns.set_theme(color_codes = True)    # set函数实现风格设置 (背景色、风格、字型、字体等)。 color_codes=True启用颜色代码映射功能。
                                               # 当设置为 True 时，可以使用简写的颜色名称（如 'b' 表示蓝色、'g' 表示绿色、'r' 表示红色）来引用调色板中对应的颜色

# 设置Matplotlib正常显示中文和负号
plt.rcParams['font.sans-serif'] = ['SimHei']  # 用黑体显示中文
plt.rcParams['axes.unicode_minus'] = False    # 正常显示负号

fig, axes = plt.subplots(1, 2, figsize=(12, 7))
sns.distplot(df_iris['Petal_Width'],ax=axes[0])
axes[0].set_title('distplot')
       # 'Petal_Width'（花瓣宽度）这一列绘制分布图
sns.histplot(df_iris['Petal_Width'], kde=True, bins=5, ax=axes[1])
axes[1].set_title('histplot')
plt.suptitle("iris数据Petal_Width的分布图");
#plt.savefig('c:/image/iris数据集中Petal_Width的分布图.png')
```

## 【例2】distplot方法绘制的直方图。

```python
sns.distplot(df_iris['Petal_Width'],bins = 30,kde = False,rug = True);
```

## 【例3】直接绘制密度图。

```python
sns.distplot(df_iris['Petal_Width'],hist = False,rug = True);
```

## 【例4】用kdeplot绘制密度图。

```python
fig,axes = plt.subplots(1,2)
sns.distplot(df_iris['Petal_Length'], ax = axes[0], kde = True, rug = True)  # kde密度曲线，rug边际毛毯
sns.kdeplot(df_iris['Petal_Length'], ax = axes[1], shade =True);             # shade阴影
```

## 【例5】distplot子图绘图。

```python
import numpy as np
sns.set(palette = "muted",color_codes = True)

f,axes = plt.subplots(2, 2, figsize = (7, 7), sharex = True)
sns.distplot(df_iris['Petal_Length'], kde = False, color = "b", ax = axes[0,0])
sns.distplot(df_iris['Petal_Length'], hist = False, rug = True, color = "r", ax = axes[0,1])
sns.distplot(df_iris['Petal_Length'], hist = False,color = "g", kde_kws = {"shade":True}, ax = axes[1,0])
sns.distplot(df_iris['Petal_Length'], color = "m", ax = axes[1,1]);

df_iris
```

花瓣长度直方图与密度图：<br>
Setosa：花瓣长度集中在 1.0–2.0 cm，分布范围极窄且对称，表明样本间差异极小。<br>
Versicolor：分布在 3.0–5.0 cm，近似正态分布，但存在轻微左偏。<br>
Virginica：分布在 4.5–7.0 cm，分布范围最广且右偏（曲线右边尾巴长，且顶峰偏左），尾部延伸至7.0 cm，可能存在少量大值样本。

## 【例6】用boxplot绘制箱线图。

```python
sns.boxplot(x=df_iris['Species'],y=df_iris['Petal_Width']);  
```

### 不同品种间差异：
setosa的petal_width（花瓣宽度）最小（均值约0.25cm），

versicolor次之（均值约1.33cm），

virginica最大（均值约2.03cm）。

分布形态：
setosa分布集中且对称，virginica右偏，versicolor接近正态分布。

## Seaborn可视化实践——小费实例

```python
fdata=pd.read_excel('./data/tips.xls')
fdata
```

```python
sns.violinplot(x=fdata['day'], y =fdata['total_bill'], palette='Accent');
#sns.violinplot(x='day', y='total_bill', data=fdata, palette='Accent')   # 第2种写法
```

上图为典型右偏分布

```python
sns.boxplot(x='day',y='total_bill', data=fdata, palette='Accent');
```

上图显示：周六和周日消费比周四和周五要高。#voilinplot和boxplot很相似，但voilinplot结合了boxplot图和密度痕迹。

```python
sns.boxplot(x='day',y ='total_bill', data=fdata, hue='sex',palette='Accent'); # 按性别（sex）分组
```

在上面的图表中可以看到，在周六时，女性买单的次数会比男性多。

```python
sns.violinplot(x='day',y='total_bill', data=fdata, hue='sex',palette='Accent'); 
```

```python
# 小提琴内部显示“细条”（stick），将 hue 分组的两类（如男/女）分别绘制在小提琴中轴线的两侧，便于直接对比
sns.violinplot(x='day',y='tip', data=fdata, hue='sex', inner='stick',split=True, palette='Accent');
```

顾客在餐厅的小费金额主要是在1-6的范围内分布的。
