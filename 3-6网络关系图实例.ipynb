```python
from pyecharts import options as opts
from pyecharts.charts import Sankey

nodes = [                      #定义节点
    {"name": "category1"},
    {"name": "category2"},
    {"name": "category3"},
    {"name": "category4"},
    {"name": "category5"},
    {"name": "category6"},
]

links = [                        #定义节点中间的链接，即关系
    {"source": "category1", "target": "category2", "value": 10},
    {"source": "category2", "target": "category3", "value": 15},
    {"source": "category3", "target": "category4", "value": 20},
    {"source": "category5", "target": "category6", "value": 25},
]
c = (
    Sankey()
    .add(
        "sankey",
        nodes,
        links,
        linestyle_opt=opts.LineStyleOpts(
            opacity=0.2,   # 线条透明度
            curve=0.5,     # 线条弯曲程度
            color="source"),   # 线条颜色跟随源节点的颜色
        label_opts=opts.LabelOpts(position="right"),  # 标签显示在右边
    )
    .set_global_opts(title_opts=opts.TitleOpts(title="Sankey-基本示例"))
)

display_chart(c)
```

```python
import pandas as pd
from pyecharts.charts import Sankey
from pyecharts import options as opts

# 读取Excel文件
data = pd.read_excel(r'data/我的生活支出一览.xlsx')

# 假设Excel文件中有三列，列名分别是'支出类别'，'支出项目'和'金额'
category_col = '支出类别'
item_col = '支出项目'
value_col = '金额'

# 生成nodes
nodes = [{'name': '总支出'}]
for category in data[category_col].unique():
    nodes.append({'name': category})

# 生成links
links = []
for index, row in data.iterrows():
    links.append({
        'source': row[category_col],
        'target': row[item_col],
        'value': row[value_col]
    })
    
# 创建桑基图
sankey = (
    Sankey()
    .add(
        "费用/元",
        nodes,
        links,
        linestyle_opt=opts.LineStyleOpts(opacity=0.2, curve=0.5, color="source", type_="dotted"),
        label_opts=opts.LabelOpts(position="right"),
    )
    .set_global_opts(title_opts=opts.TitleOpts(title="我的生活支出一览"))
)


display_chart(sankey)
```

## 3.6.2 网络图

```python
from pyecharts import options as opts
from pyecharts.charts import Graph

nodes_data = [
    opts.GraphNode(name="结点1", symbol_size=10),
    opts.GraphNode(name="结点2", symbol_size=20),
    opts.GraphNode(name="结点3", symbol_size=30),
    opts.GraphNode(name="结点4", symbol_size=40),
    opts.GraphNode(name="结点5", symbol_size=50),
    opts.GraphNode(name="结点6", symbol_size=60),
]
links_data = [
    opts.GraphLink(source="结点1", target="结点2", value=2),
    opts.GraphLink(source="结点2", target="结点3", value=3),
    opts.GraphLink(source="结点3", target="结点4", value=4),
    opts.GraphLink(source="结点4", target="结点5", value=5),
    opts.GraphLink(source="结点5", target="结点6", value=6),
    opts.GraphLink(source="结点6", target="结点1", value=7),
]
c = (
    Graph()
    .add(
        "",
        nodes_data,
        links_data,
        repulsion=4000,   #repulsion 参数设置了节点之间的斥力
        edge_label=opts.LabelOpts(
            is_show=True, position="middle", formatter="{b} 的数据 {c}"   #edge_label 设置了边的标签样式和显示位置
        ),
    )
    .set_global_opts(
        title_opts=opts.TitleOpts(title="Graph-GraphNode-GraphLink-WithEdgeLabel")
    )
    #.render("graph_with_edge_options.html")
)

display_chart(c)
```

## 网络图实例 海绵宝宝动画片角色关系图

```python
from pyecharts.charts import Graph

nodes_data = [
    opts.GraphNode(name="海绵宝宝", symbol_size=80  ),
    opts.GraphNode(name="派大星", symbol_size=45  ),
    opts.GraphNode(name="小蜗", symbol_size=45   ),
    opts.GraphNode(name="蟹老板", symbol_size=60   ),
    opts.GraphNode(name="章鱼哥", symbol_size=45   ),
    opts.GraphNode(name="泡芙老师", symbol_size=60  ),
    opts.GraphNode(name="痞老板", symbol_size=60   ),
    opts.GraphNode(name="拉里", symbol_size=45    ),
    opts.GraphNode(name="射线恶魔", symbol_size=50  ),
    opts.GraphNode(name="珍珍", symbol_size=45   ),

]
links_data = [
    opts.GraphLink(source="海绵宝宝", target="派大星", value='好友',symbol = ['arrow','arrow']),
    opts.GraphLink(source="海绵宝宝", target="小蜗", value='好友',symbol = ['arrow','arrow']),
    opts.GraphLink(source="海绵宝宝", target="章鱼哥", value='邻居',symbol = ['arrow','arrow']),
    opts.GraphLink(source="海绵宝宝", target="蟹老板", value='雇佣',symbol = ['arrow',None]),
    opts.GraphLink(source="海绵宝宝", target="泡芙老师", value='师徒',symbol = ['arrow','arrow']),
    opts.GraphLink(source="海绵宝宝", target="痞老板", value='敌人',symbol = ['arrow','arrow']),
    opts.GraphLink(source="蟹老板", target="章鱼哥", value='雇佣',symbol = [None,'arrow']),
    opts.GraphLink(source="蟹老板", target="痞老板", value='敌人',symbol = ['arrow','arrow']),
    opts.GraphLink(source="海绵宝宝", target="拉里", value='崇拜',symbol=[None,'arrow']),
    opts.GraphLink(source="海绵宝宝", target="射线恶魔", value='亦敌亦友',symbol = ['arrow','arrow']),
    opts.GraphLink(source="海绵宝宝", target="珍珍", value='好友',symbol = ['arrow','arrow']),
    opts.GraphLink(source="蟹老板", target="珍珍", value='父女',symbol=[None,'arrow']),
]
c = (
    Graph(init_opts=opts.InitOpts(width='1470px',height='730px'))
    .add(
        "",
        nodes_data,
        links = links_data,
        repulsion=5000,       #repulsion 参数设置了节点之间的斥力
        is_draggable = True,     #is_draggable 允许用户拖动节点。
        edge_label=opts.LabelOpts(
            is_show=True, position="middle", formatter="关系:{c}",                #edge_label 设置了边的标签样式和显示位置
        ),
        label_opts = opts.LabelOpts(color='white',font_size=12)
    )
    .set_global_opts(title_opts=opts.TitleOpts(title="《海绵宝宝》动画片角色关系图"))
)

display_chart(c)
```

## 环形图实例-《悲惨世界》人物关系图
该图基于维克多·雨果的小说《悲惨世界》中的人物关系数据，展示了小说中各个角色之间的联系，例如谁与谁相遇过，他们之间的关系如何。

```python
import json
from pyecharts import options as opts
from pyecharts.charts import Graph

with open("data/les-miserables.json", "r", encoding="utf-8") as f:
    j = json.load(f)
    nodes = j["nodes"]
    links = j["links"]
    categories = j["categories"]
print(nodes[:2])
print(links[:5])
print(categories)

c = (
    Graph(init_opts=opts.InitOpts(width="1000px", height="600px"))
    .add(
        "",
        nodes=nodes,
        links=links,
        categories=categories,
        layout="circular",          #提供了其他几种布局选项，如 "circular"（环形）、"none"（无布局，自定义位置）、"force"（力导向布局）
        is_rotate_label=True,    #is_rotate_label 设置为 True 表示标签旋转
        linestyle_opts=opts.LineStyleOpts(color="source", curve=0.3),  #设置了边的样式。
        label_opts=opts.LabelOpts(position="right"),                   #设置了标签的样式。
    )
    .set_global_opts(
        title_opts=opts.TitleOpts(title="Graph-Les Miserables"),
        legend_opts=opts.LegendOpts(orient="vertical", pos_left="2%", pos_top="20%"),
    )
)

display_chart(c)
```
