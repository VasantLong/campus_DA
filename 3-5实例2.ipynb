## 实例1 全国主要城市航班流动图为例（虚拟数据）

```python
from pyecharts import options as opts 
from pyecharts.charts import Geo 
from pyecharts.globals import ChartType, SymbolType

c = (
    Geo()
    .add_schema(maptype="china")
    .add( "",
        [ ("哈尔滨", 66), ("重庆", 88), ("上海", 100), ("乌鲁木齐", 30),("北京", 30),("兰州",170)],
        type_=ChartType.EFFECT_SCATTER,
        color="green",
    )
    .add(
        "geo",
        [("北京", "兰州"),( "兰州","北京"), ("重庆", "上海"),("哈尔滨", "重庆"),("乌鲁木齐", "哈尔滨")],
        type_=ChartType.LINES,
        effect_opts=opts.EffectOpts(symbol=SymbolType.ARROW, symbol_size=6, color="blue"),# 设置特效图形为箭头，特效图形的大小、颜色
        linestyle_opts=opts.LineStyleOpts(curve=0.2),  #设置线条样式 curve 参数设置为 0.2，表示线条将有轻微的曲线效果。
    )
    .set_series_opts(label_opts=opts.LabelOpts(is_show=False))
    .set_global_opts(title_opts=opts.TitleOpts(title="主要城市航班路线和数量"))
)

display_chart(c)
```

```python
import re
import pandas as pd
from pyecharts.charts import Map
from pyecharts import options as opts 

data = pd.read_excel('c:/dataset/03-07.xls')
city = data.iloc[5:26,0].apply(lambda x: re.sub(r'\s+', '', x) +'市') # re.sub(r'\s+', '', x) —— 使用正则表达式 \s+ 匹配一个或多个空白字符（空格、制表符、换行等），将它们替换为空字符串，即去除所有空白。
population = data.iloc[5:26,11]
z3 = [list(i) for i in zip(city,population)]
z3
```

```python
map=(
    Map(init_opts=opts.InitOpts(width="1200px",height='600px'))
    .add('广东人口', data_pair=z3, maptype='广东', is_map_symbol_show=False)
    .set_series_opts(label_opts=opts.LabelOpts(is_show=True))
    .set_global_opts(title_opts=opts.TitleOpts(title='广东人口分布图', subtitle='数据来源：广东统计年鉴'),
                     visualmap_opts=opts.VisualMapOpts(max_=2000,is_piecewise=True,
                                                       range_color=['lightskyblue','yellow','orangered'])
                    )
)
```

```python
map.load_javascript()
```

```python
map.render_notebook()
```

```python

geo=(
    Geo(init_opts=opts.InitOpts(width="1200px",height='600px'))
    .add_schema(maptype='广东',itemstyle_opts=opts.ItemStyleOpts(color='#333333',border_color='#FFFF22'))  # 设置项目样式
    .add('广东人口',z3,label_opts=opts.LabelOpts(is_show=False),type_=ChartType.EFFECT_SCATTER)
    .set_global_opts(title_opts=opts.TitleOpts(title='广东人口分布图',subtitle='数据来源：广东统计年鉴'),
                     visualmap_opts=opts.VisualMapOpts(max_=2000,is_piecewise=True,
                                                       range_color=['lightskyblue','yellow','orangered'])
                    )
)
geo.load_javascript()
```

```python
geo.render_notebook()
```

```python
import pandas as pd
from pyecharts.charts import Map
from pyecharts import options as opts 
from pyecharts.globals import ChartType

data = pd.read_excel('c:/dataset/02-06.xls',sheet_name='Sheet1')
city = data.iloc[9:-1,0].str.strip(' ')  # 对 Series 中的每个字符串元素：strip(' ') 会删除字符串开头和结尾的空格字符，注意：它不会删除字符串中间的空格，也不会删除其他空白字符（如 \t、\n）。
population = data.iloc[9:-1,3]
z3 = [list(i) for i in zip(city,population)]

data
```

```python
map=(
    Map(init_opts=opts.InitOpts(width="1200px",height='600px'))
    .add('天津人口',data_pair=z3,maptype='天津',is_map_symbol_show=False)
    .set_series_opts(label_opts=opts.LabelOpts(is_show=True))
    .set_global_opts(title_opts=opts.TitleOpts(title='天津人口分布图',subtitle='数据来源：天津统计年鉴'),
                     visualmap_opts=opts.VisualMapOpts(max_=210,is_piecewise=True,
                                                       range_color=['lightskyblue','yellow','orangered'])
                    )
)
map.load_javascript()
```

```python
map.render_notebook()
```
