```python
import pyecharts.options as opts
from pyecharts.charts import Line
import pandas as pd
# import xlrd

#导入数据
df = pd.read_excel('./data/居民人均收支.xls',header=0)
df
```

```python

x_data = ['2016','2017','2018','2019','2020']
y1_data = list(df.iloc[0,5:0:-1])
y2_data = list(df.iloc[6,5:0:-1])

```

```python
#获取数据
x_data = ['2016年','2017年','2018年','2019年','2020年']
#获取居民人均可支配收入数据
data1 = list(df.iloc[0,[5,4,3,2,1]])
#获取居民人均消费支出数据
data2 = list(df.iloc[6,[5,4,3,2,1]])
print(data1)
print(data2)

y1_data = data1
y2_data = data2

#绘图折线图
line=(
    Line()
     .add_xaxis(xaxis_data=x_data)
    .add_yaxis(
        series_name="居民人均可支配收入",
        y_axis=y1_data,
        symbol="circle",
        is_symbol_show=True,
        label_opts=opts.LabelOpts(is_show=True),
    )
    .add_yaxis(
        series_name="居民人均消费支出",
        y_axis=y2_data,
        symbol="emptyCircle",
        is_symbol_show=True,
        label_opts=opts.LabelOpts(is_show=True)
    )
    .set_global_opts(
        tooltip_opts=opts.TooltipOpts(is_show=False),    # 不显示提示框
        xaxis_opts=opts.AxisOpts(name='年份', type_="category"),  # X轴标签配置，类型为类别轴，适用于离散的文本/类别数据（如年份、月份）
        yaxis_opts=opts.AxisOpts(
            name='金额', type_="value",     # 坐标轴类型为数值轴，适用于连续数值数据
            axistick_opts=opts.AxisTickOpts(is_show=True)   # 显示坐标轴刻度线（即轴线上的小短线）
        ),
        title_opts=opts.TitleOpts(
            title="2016—2020年我国居民人均收支",
                subtitle="单位：元",pos_left='center'),
        #添加脚注
        graphic_opts=opts.GraphicGroup(
            graphic_item=opts.GraphicItem(left='12%',bottom='0%'),
            children=[
                opts.GraphicText(graphic_textstyle_opts=
                opts.GraphicTextStyleOpts(text='数据来源：国家统计局',
                                          font="14px Microsoft YaHei")
            )]
        ),
        legend_opts=opts.LegendOpts(pos_right='right',orient='vertical')
    )
   
    #.render("c:/html/task6-23.html")   
 )
display_chart(line)
```

```python
import pyecharts.options as opts
from pyecharts.charts import Bar
from pyecharts.charts import Line
import pandas as pd
import xlrd

#导入数据
df = pd.read_excel('./data/城市绿地面积.xls',header=0)
df
```

```python
#获取城市绿地面积和建成区绿化覆盖率（%）数据
df1 = df.iloc[0,[5,4,3,2,1]]
data1 = list(df1)
print(data1)
data2 = list(df.iloc[4,[5,4,3,2,1]])
print(data2)
lable = list(df1.index)
print(lable)

#绘制单数据系列柱形图
bar = (
    Bar()
    .add_xaxis(lable)
    .add_yaxis("城市绿地面积", data1,bar_width=70)
    .set_global_opts(title_opts=opts.TitleOpts(
        title="2015—2019年城市绿地面积",
        subtitle="单位：万公顷"),
        #添加脚注
        graphic_opts=opts.GraphicGroup(
           graphic_item=opts.GraphicItem(left='12%',bottom='0%'),
           children=[
               opts.GraphicText(graphic_textstyle_opts=
               opts.GraphicTextStyleOpts(text='数据来源：国家统计局',
                                         )
           )
           ]
        )
    )
    #在右边新增y坐标轴
    .extend_axis(
        yaxis=opts.AxisOpts(     # y轴配置
            name="建成区绿化覆盖率（%）",
            type_="value",
            min_=0,
            max_=43,
            position="right",
            #设置坐标轴刻度线配置项
            axisline_opts=opts.AxisLineOpts(
                #设置坐标轴线为红色
                linestyle_opts=opts.LineStyleOpts(color="red")
            ),
            #设置坐标轴上标签内容格式器
            axislabel_opts=opts.LabelOpts(formatter="{value} %"),
       )
    )
)

#绘制折线图
line = (
    Line()
    .add_xaxis(xaxis_data=lable)
    .add_yaxis(
         series_name="建成区绿化覆盖率",
         y_axis=data2,
         yaxis_index=1,  # 当图表包含多个 Y 轴时，指定该系列绑定到哪个 Y 轴（0: 主Y轴，1: 第一个次Y轴，依此类推）。
         color="#675bba",
         label_opts=opts.LabelOpts(is_show=True),
     )
)
#将折线图叠加到柱形图上
bar.overlap(line)
#bar.render("d:/html/task6-25.html")

display_chart(bar)
```
